## Coordinate Frames

Listed in order from the most user facing, to the back end internal raw data.

<div style="text-align:center;">

### 2D UI Frames

| Frame     | Units         | Origin           | Y dir | Description |
| -----     | ------------- | ---------------- | ----- | ----------- |
| qwn       | window pixels | window top left  | down  | 2D Qt window coordinates |
| ndc       | half windows  | viewport center  | up    | 2D OpenGL viewport normalized device coordinates |
| nic (prj) | half images   | viewport center  | up    | 2D normalized viewport coordinates corrected for scale and aspect |

### 2D Raster Image Frames

| Frame     | Units         | Origin           | Y dir | Description |
| -----     | ------------- | ---------------- | ----- | ----------- |
| opx (omp) | image pixels  | image top left   | down  | 2D logical image pixels considering EXIF orientation |
| rpx (rmp) | image pixels  | image top left   | down  | 2D internal image pixels before EXIF orientation correction |
| otc (tco) | images        | image bottom left   | up  | 2D full image logical texture coordinates considering exif orientation |
| rtc (tcr) | images        | image bottom left   | up  | 2D full image true texture coordinates |
| ttc (tct) | images        | tile bottom left    | up  | 2D tile logical texture coordinates, as if unpadded |
| ptc       | images        | tile bottom left    | up  | 2D tile true texture coordinates including padding on the edges|

### 360 Panorama Frames

| Frame     | Units         | Origin           | Y dir | Description |
| -----     | ------------- | ---------------- | ----- | ----------- |
| usr (obq) | radians       | user eye         | top of head | 3D app user's view frame|
| geo (ont) | radians       | photographer's camera  | sky   | 3D real world north-forward frame centered on camera |
| pcm (raw) | radians       | photographer's camera  | camera top | 3D X-right Z-into-lens direction camera frame |
| eqr       | radians       | image center     | up    | 2D heading, pitch (equirectangular projection) |

### VR Frames
| Frame     | Units         | Origin           | Y dir | Description |
| -----     | ------------- | ---------------- | ----- | ----------- |
| hmd       | meters        | user eye         | top of head | 3D HMD user's eye view frame |
| stg       | meters        | fixed point in user's room | ceiling | 3D real world room frame |
| spx       | texels        | lower left       | up    | 2D swapchain texel coordinates |
| stc       | images        | lower left       | up    | 2D swapchain texture coordinates |
| cnv       | images        | canvas center    | up    | 2D virtual canvas texture coordinates |

</div>

Standard photo rendering goes qwn->ndc->nic->opx->otc->rtc->ttc->ptc

Panoramic photo rendering goes qwn->ndc->nic->usr->geo->pcm->otc->rtc->ttc->ptc

## Transform parameters

1. Zoom: windows per image
2. Direction: heading, pitch degrees (360 only)
3. Image center: image pixel x, y (rectangular only)
4. Image width, height: image pixels (respecting EXIF orientation)
5. Window width, height: screen pixels

### Situations where transformation between coordinate frames is needed:

 1. When single dragging images to pan. Here we need the jacobian matrix relating qwn coordinates to view_model parameters. $viw_J_qwn$
     * image center (but not zoom), in the case of rectangular images
     * view pitch and heading angles (but not zoom nor roll) in the case of 360 images
 2. When rendering images in a GLSL shader. $txc_X_ndc$
 3. When hovering with the mouse $opx_X_qwn$ for rectangular and $viw_X_qwn$ for 360 images.

The only frames we need to compute are:
  * tex, in the rectangular shader tex_X_ndc
  * opx, on hover opx_X_qwn
  * derivative of image center (in either geo, pcm, tex, img, or opx), on drag opx_J_qwn
    * ypr for 360s ypr_J_qwn
    * opx for rectangular opx_J_qwn
  
Rectangular view state:
  * image center (should be opx, TODO:)
  * zoom (windows per image)
360 view state:
  * pitch, yaw (maybe as equirect image center in degrees)
  * zoom (windows per image at center?)
  
What are normalized image coordinates (nic) ?
  * origin (0,0) is at image center
  * Y increases upward
  * EXIF reorientation is applied
  * coordinates range from -1 (left, bottom) to +1 (right, top)
  * scaled such that underlying full image exactly fits in the window. The edge touching dimension ranges from -1 to +1
  
Vertex shader converts from ndc to nic
  * apply zoom, window aspect, image size, image center
Fragment shader copxutes from nic to txc, and samples
  * apply pixel filter
  * 360 images:
    * nic to geocentric unit sphere
    * rotate by yaw pitch roll
    * project to projection...

The transform required for hover coordinates is composed of sub-transforms:

$^{opx}p \ = \ ^{opx}X^{qwn} \ \cdot \ ^{qwn}p $

$^{opx}X^{qwn} \ = \ ^{opx}X^{nic} \ \cdot \ ^{nic}X^{ndc} \ \cdot \ ^{ndc}X^{qwn}$


```python
# printed symbols of transforms
opx_X_qwn_s = symbols("^{opx}X^{qwn}")
ndc_X_qwn_s = symbols("^{ndc}X^{qwn}")
nic_X_ndc_s = symbols("^{nic}X^{ndc}")
opx_X_nic_s = symbols("^{opx}X^{nic}")
```

```python
# ndc_X_qwn: Transform to OpenGL normalized device coordinates from Qt window pixel coordinates

# window dimensions
w_win, h_win = symbols("w_win, h_win")

ndc_X_qwn_m = Matrix([
    [2/w_win, 0, -1],
    [0, -2/h_win, 1],
    [0, 0, 1],
])

# confidence checks

# Upper left corner
ulc_qwn_v = Matrix([0, 0, 1])
ulc_ndc_v = Matrix([-1, 1, 1])
assert ndc_X_qwn_m * ulc_qwn_v == ulc_ndc_v

# Upper right corner
urc_qwn_v = Matrix([w_win, 0, 1])
urc_ndc_v = Matrix([1, 1, 1])
assert ndc_X_qwn_m * urc_qwn_v == urc_ndc_v

# Lower left corner
llc_qwn_v = Matrix([0, h_win, 1])
llc_ndc_v = Matrix([-1, -1, 1])
assert ndc_X_qwn_m * llc_qwn_v == llc_ndc_v

# Lower right corner
lrc_qwn_v = Matrix([w_win, h_win, 1])
lrc_ndc_v = Matrix([1, -1, 1])
assert ndc_X_qwn_m * lrc_qwn_v == lrc_ndc_v

Eq(ndc_X_qwn_s, ndc_X_qwn_m, evaluate=False)
```

```python
Eq(symbols("^{qwn}X^{ndc}"), ndc_X_qwn_m.inv(), evaluate=False)
```

```python
# nic_X_ndc: transform to vimage normalized image coordinates from OpenGL normalized device coordinates

w_omp, h_omp = symbols("w_omp, h_omp")  # (reoriented) image dimensions
zoom = symbols("zoom")  # windows per image (default 1)
cen_x_omp, cen_y_omp = symbols("^{omp}cen_x ^{omp}cen_y")

# TODO: aspect, center
aspect_omp = w_omp / h_omp
aspect_qwn = w_win / h_win

# Case A: If aspect_omp > aspect_qwn, image is fatter than window, so pad top/bottom at zoom == 1
# Case B: otherwise, pad left/right at zoom == 1

aspect_scale_ompA = w_omp
aspect_scale_ompB = h_omp
aspect_scale_qwnA = w_win
aspect_scale_qwnB = h_win

# Create a symbol for the A/B choice of aspect scaling
asc_omp, asc_qwn = symbols("asc_omp asc_qwn")

aspect_scale = Matrix([
    [w_win / asc_qwn, 0, 0],
    [0, h_win / asc_qwn, 0],
    [0, 0, 1],
])

zoom_scale = Matrix([
    [1 / zoom, 0, 0],
    [0, 1 / zoom, 0],
    [0, 0, 1],
])


nic_X_ndc_m = zoom_scale * aspect_scale

Eq(nic_X_ndc_s, nic_X_ndc_m, evaluate=False)
```

```python
# opx_X_nic: Transform from normalized image coordinates to oriented image pixels

w_opx, h_opx = symbols("w_opx, h_opx")
cen_x_opx, cen_y_opx = symbols("cen_x_opx, cen_y_opx")
asc_opx = symbols("asc_opx")

center_shift = Matrix([
    [1, 0, -2 * (w_opx / 2 - cen_x_opx) / asc_opx],
    [0, 1, 2 * (h_opx / 2 - cen_y_opx) / asc_opx],
    [0, 0, 1],
])

opx_X_nic_m = Matrix([
    [asc_opx / 2, 0, w_opx / 2],
    [0, -asc_opx / 2, h_opx / 2],
    [0, 0, 1],
]) * center_shift

Eq(opx_X_nic_s, opx_X_nic_m, evaluate=False)
```

```python
Eq(symbols("^{opx}X^{ndc}"), opx_X_nic_m * nic_X_ndc_m, evaluate=False)
```

```python
Eq(symbols("^{ndc}X^{opx}"), (opx_X_nic_m * nic_X_ndc_m).inv(), evaluate=False)
```

```python
opx_X_qwn_m = opx_X_nic_m * nic_X_ndc_m * ndc_X_qwn_m
Eq(opx_X_qwn_s, opx_X_qwn_m, evaluate=False)
```

```python
nic_X_qwn_s = symbols("^{nic}X^{qwn}")
Eq(nic_X_qwn_s, nic_X_ndc_m * ndc_X_qwn_m, evaluate=False)
```

```python
# Confidence checks

# Test case for zoom factor 1
test_m = opx_X_qwn_m.subs(zoom, 1)
# Default image center
test_m = test_m.subs(cen_x_opx, w_opx/2).subs(cen_y_opx, h_opx/2)
# Square window, square image
test_m = test_m.subs(h_opx, w_opx).subs(h_win, w_win).subs(asc_opx, w_opx).subs(asc_qwn, w_win)
assert test_m * Matrix([0, 0, 1]) == Matrix([0, 0, 1])  # Upper left corner
assert test_m * Matrix([w_win, w_win, 1]) == Matrix([w_opx, w_opx, 1])  # Lower right corner
assert test_m * Matrix([w_win / 2, w_win / 2, 1]) == Matrix([w_opx / 2, w_opx / 2, 1])  # Center

# Test case for zoom factor 2
test_m = opx_X_qwn_m.subs(zoom, 2)
# Default image center
test_m = test_m.subs(cen_x_opx, w_opx/2).subs(cen_y_opx, h_opx/2)
# Square window, square image
test_m = test_m.subs(h_opx, w_opx).subs(h_win, w_win).subs(asc_opx, w_opx).subs(asc_qwn, w_win)
assert test_m * Matrix([w_win / 2, w_win / 2, 1]) == Matrix([w_opx / 2, w_opx / 2, 1])  # Center unchanged
assert test_m * Matrix([0, 0, 1]) == Matrix([w_opx / 4, w_opx / 4, 1])  # Upper left is offset by 1/4

# Test case for center at upper left
test_m = opx_X_qwn_m.subs(zoom, 1)
# image center at upper left
test_m = test_m.subs(cen_x_opx, 0).subs(cen_y_opx, 0)
# Square window, square image
test_m = test_m.subs(h_opx, w_opx).subs(h_win, w_win).subs(asc_opx, w_opx).subs(asc_qwn, w_win)
assert test_m * Matrix([0, 0, 1]) == Matrix([-w_opx/2, -w_opx/2, 1]) # Upper left of window is in negative image coordinates
assert test_m * Matrix([w_win, w_win, 1]) == Matrix([w_opx/2, w_opx/2, 1]) # Lower right of window is in image center
assert test_m * Matrix([w_win/2, w_win/2, 1]) == Matrix([0, 0, 1])  # Window center is image center

# Default image center
test_m = opx_X_qwn_m.subs(cen_x_opx, w_opx/2).subs(cen_y_opx, h_opx/2)
assert test_m * Matrix([w_win/2, h_win/2, 1]) == Matrix([w_opx/2, h_opx/2, 1]) # Window center is image center
```

```python
# Conversion from image pixel coordinates to oriented texture coordinates

otc_X_opx = Matrix([
    [1/w_opx, 0, 0],
    [0, 1/h_opx, 0],
    [0, 0, 1],
])

Eq(symbols("^{otc}X^{opx}"), otc_X_opx, evaluate=False)
```

```python
# Jacobian for dragging in rectangular images
d_opx_x, d_opx_y = symbols("{\partial}^{opx}p_x, {\partial}^{opx}p_y")
d_qwn_x, d_qwn_y = symbols("{\partial}^{qwn}p_x, {\partial}^{qwn}p_y")
ue = UnevaluatedExpr
Eq(Matrix([d_opx_x, d_opx_y]), ue(opx_X_qwn_m[0:2,0:2]) * ue(Matrix([d_qwn_x, d_qwn_y]).T), evaluate=False)
```

```python
# Final expressions for the change in image pixels for a change in window pixels.
# Increment the image center by the negative of this change, to pan the image with the mouse.
Eq(d_opx_x, opx_X_qwn_m[0,0] * d_qwn_x, evaluate=False), Eq(d_opx_y, opx_X_qwn_m[1,1] * d_qwn_y, evaluate=False)
```

```python
# For 360 images, we want to pause at the nic coordinates, 
#  then add projection-specific transforms for each 360 projection
nic_X_qwn_s = symbols("^{nic}X^{qwn}")
nic_X_qwn_m = nic_X_ndc_m * ndc_X_qwn_m
Eq(nic_X_qwn_s, nic_X_qwn_m, evaluate=False)
```

```python
# Projected 360 prj coordinates differ from nic by an isotropic scale
prj_X_nic_s = symbols("^{prj}X^{nic}")
prj_X_nic_m = Matrix([
    [pi/2, 0, 0],
    [0, pi/2, 0],
    [0, 0, 1],
])
Eq(prj_X_nic_s, prj_X_nic_m, evaluate=False)
```

```python
prj_X_qwn_s = symbols("^{prj}X^{nic}")
prj_X_qwn_m = prj_X_nic_m * nic_X_qwn_m
Eq(prj_X_qwn_s, prj_X_qwn_m, evaluate=False)
```

```python
# Partial Jacobian matrix for dragging 360 images
nic_J_qwn_s = symbols("^{nic}J^{qwn}")
nic_J_qwn_m = nic_X_qwn_m[0:2,0:2]  # Just the scale, because it's linear or whatever.
Eq(nic_J_qwn_s, nic_J_qwn_m, evaluate=False)
```

```python
# Projected unzoomed coordinates for 360 are scaled to +- pi/2 on screen
prj_J_qwn_s = symbols("^{prj}J^{qwn}")
prj_J_qwn_m = nic_J_qwn_m * pi / 2
prj_J_qwn_m
Eq(prj_J_qwn_s, prj_J_qwn_m, evaluate=False)
```

```python
# Equirectangular coordinates - heading and pitch (in radians)
# heading is radians clockwise from north/straight-ahead when viewed from above
# pitch is radians above the horizon

# usr is xyz unit sphere coordinates "oblique" because a view transform may have been applied already
# right-handed x-positive-right, y-positive-up, z-positive-nearer
x_usr, y_usr, z_usr = symbols("^{usr}p_x, ^{usr}p_y, ^{usr}p_z")
p_usr_s = symbols("^{usr}p")

eqr_from_usr = Matrix([
    atan2(x_usr, -z_usr),  # negative Z, because the central pixel is at the far end of the sphere
    asin(y_usr),
])

p_eqr_s = symbols("^{eqr}p")
Eq(p_eqr_s, eqr_from_usr, evaluate=False)
```

```python
eqr_x, eqr_y = symbols("^{eqr}p_x, ^{eqr}p_y")
usr_from_eqr = Matrix([
    [cos(eqr_y) * sin(eqr_x)],
    [sin(eqr_y)],
    [-cos(eqr_y) * cos(eqr_x)],
])

Eq(p_usr_s, usr_from_eqr, evaluate=False)
```

```python
# Stereographic coordinates
# Assume that ste coordinates range from -1 to +1 at zoom==1

# 2D Stereographic image coordinates. Range is unlimited.
ste_x, ste_y = symbols("^{ste}p_x, ^{ste}p_y")

d = ste_x**2 + ste_y**2 + 4
usr_from_ste = Matrix([
    [4 * ste_x / d],
    [4 * ste_y / d],
    [(d - 8) / d],
])

Eq(p_usr_s, usr_from_ste, evaluate=False)
```

```python
# Stereographic Jacobian usr_J_ste
usr_J_ste_m = Matrix([
    [diff(usr_from_ste[0], ste_x), diff(usr_from_ste[0], ste_y)],
    [diff(usr_from_ste[1], ste_x), diff(usr_from_ste[1], ste_y)],
    [diff(usr_from_ste[2], ste_x), diff(usr_from_ste[2], ste_y)],
])
dp_usr_x, dp_usr_y, dp_usr_z = symbols("^{usr}{\partial}p_x ^{usr}{\partial}p_y ^{usr}{\partial}p_z")
dp_ste_x, dp_ste_y = symbols("^{ste}{\partial}p_x ^{ste}{\partial}p_y")
dp_usr = Matrix([dp_usr_x, dp_usr_y, dp_usr_z])
dp_ste = Matrix([dp_ste_x, dp_ste_y])
Eq(dp_usr, ue(usr_J_ste_m) * ue(dp_ste.transpose()), evaluate=False)
```

## Pixels per texel evaluation using Jacobians

Fragment shader derivatives are not working well for tile boundaries, nor for texel-to-pixel scaling near tile boundaries.
So we want to compute analytic Jacobians to help with such calculattions.

    * we want qwn_J_rpx, applied to logical image texture coordinate
    * qwn_J_rpx = (qwn_J_ndc * ndc_J_nic) * nic_J_usr * (usr_J_geo * geo_J_pcm) * pcm_J_opx * opx_J_rpx
      a) qwn_J_nic can be computed on the host
      b) nic_J_usr depends on the display projection and the screen location, so needs to be computed in the fragment shader
      c) usr_J_pcm can be computed on the host
      d) pcm_J_opx depends on the input format and the screen location, so needs to be computed in the fragment shader
      e) opx_J_rpx can be computed on the host


### a) compute *qwn_J_nic* on host side

the change in window pixels per change in normalized image coordinates

```python
qwn_X_ndc_m = ndc_X_qwn_m.inv()
qwn_X_ndc = symbols("^{qwn}X^{ndc}")
Eq(qwn_X_ndc, qwn_X_ndc_m, evaluate=False)
```

```python
qwn_J_ndc_m = qwn_X_ndc_m[0:2,0:2]
qwn_J_ndc = symbols("^{qwn}J^{ndc}")
Eq(qwn_J_ndc, qwn_J_ndc_m, evaluate=False)
```

```python
ndc_X_nic_m = nic_X_ndc_m.inv()
ndc_X_nic = symbols("^{ndc}X^{nic}")
Eq(ndc_X_nic, ndc_X_nic_m, evaluate=False)
```

```python
ndc_J_nic_m = ndc_X_nic_m[0:2,0:2]
ndc_J_nic = symbols("^{ndc}J^{nic}")
Eq(ndc_J_nic, ndc_J_nic_m, evaluate=False)
```

```python
# This is the first host-side Jacobian required
qwn_J_nic_m = qwn_J_ndc_m @ ndc_J_nic_m
qwn_J_nic = symbols("^{qwn}J^{nic}")
Eq(qwn_J_nic, qwn_J_nic_m, evaluate=False)
```

### b) compute *nic_J_usr* in fragment shader, from usr x, y, z

This Jacobian also depends on the display projection: stereographic, perspective, equirectangular, or equidistant
